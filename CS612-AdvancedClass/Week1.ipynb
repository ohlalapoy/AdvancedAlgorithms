The GCD problem หาหารร่วมมาก Find the greatest common divisor (GCD) of two nonnegative integers, not both zero.

วิธีคิด Euclid's Algorithm : gcd (a,b) = gcd(a, a mod b) 

WORKED EXAMPLE: gcd(18, 12)  
gcd(18, 12) = gcd(12, 6)  
gcd(12, 6) = gcd(6, 0)  
gcd(6, 0) = 6


```python
def gcd(m,n):
    while n != 0:
        r = m % n #ให้ r คือ เศษ จากการหาร
        m = n # เศษยังไม่ใช่ 0 ให้เลื่อนไปทำรอบถัดไปคือเอา n มาเป้นค่า m แทน
        n = r #และเลื่อน new r = n แทนเพื่อวนหา gcd ค่าต่อไปจนได้ r = 0
    return m


a = gcd(18,12)
print(a)
```

Fibonacci problem คือการบวกค่าเลขก่อนถ้า กับเลขตัวถัดไป ต่อกันไปเรื่อยๆ จนได้ 0 1 1 2 3 5 8 .....  
F(n) = F(n-1) + F(n-2) if n > 1  
Naive Recursion คือการรันบวกค่าไปโดยที่ไม่จำว่าค่าก่อนหน้าถูกรันไปแล้วทำให้ต้องรันค่านั้นซ้ำๆ  
e.g. F(5) = F(4) + F(3) ---> F(3) = F(2)+F(1) / F(4) = F(2)+F(3) จะเห้นได้ว่ามีการรัน F(2) ซ้ำกัน ถูกรันใหม่เพราะไม่จำว่าตัวเองรันไปแล้ว



```python
def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)
```
