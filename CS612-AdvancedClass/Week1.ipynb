The GCD problem หาหารร่วมมาก Find the greatest common divisor (GCD) of two nonnegative integers, not both zero.

วิธีคิด Euclid's Algorithm : gcd (a,b) = gcd(a, a mod b) 

WORKED EXAMPLE: gcd(18, 12)  
gcd(18, 12) = gcd(12, 6)  
gcd(12, 6) = gcd(6, 0)  
gcd(6, 0) = 6


```python
def gcd(m,n):
    while n != 0:
        r = m % n #ให้ r คือ เศษ จากการหาร
        m = n # เศษยังไม่ใช่ 0 ให้เลื่อนไปทำรอบถัดไปคือเอา n มาเป้นค่า m แทน
        n = r #และเลื่อน new r = n แทนเพื่อวนหา gcd ค่าต่อไปจนได้ r = 0
    return m


a = gcd(18,12)
print(a)
```

Fibonacci problem คือการบวกค่าเลขก่อนถ้า กับเลขตัวถัดไป ต่อกันไปเรื่อยๆ จนได้ 0 1 1 2 3 5 8 .....  
F(n) = F(n-1) + F(n-2) if n > 1  
Naive Recursion คือการรันบวกค่าไปโดยที่ไม่จำว่าค่าก่อนหน้าถูกรันไปแล้วทำให้ต้องรันค่านั้นซ้ำๆ  
e.g. F(5) = F(4) + F(3) ---> F(3) = F(2)+F(1) / F(4) = F(2)+F(3) จะเห้นได้ว่ามีการรัน F(2) ซ้ำกัน ถูกรันใหม่เพราะไม่จำว่าตัวเองรันไปแล้ว



```python
def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)
```

Dynamic Programming จะ เก็บคำตอบที่คำนวณแล้ว

```python
def fib(n):
    dp = [0] * (n + 1) #สร้าง List เปล่ามารอเก้บค่าผลลัพธ์ ตามจำนวน n ที่ใส่เข้ามา

    # ตัวเร่มต้น ตำแหน่งที่ 0 1 ของ List 
    dp[0] = 0 
    dp[1] = 1

    for i in range(2, n + 1): #เริ่มที่ index 2 เนื่องจาก index 0,1 คือตัวแรก 0,1 
        dp[i] = dp[i-1] + dp[i-2]

    return dp[n],dp #เมื่อยวกค่าแล้วเอาไปเก้บใน list เราวามารถเรียกค่านั้นออกมาได้เลย

a = fib(34)
print(a)
```

Sorting การจัดเรียงข้อมูล

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]          # ตัวที่จะแทรก
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]   # เลื่อนไปทางขวา
            j -= 1
        arr[j + 1] = key      # หย่อนลงตำแหน่งที่ว่าง
    return arr


def selection_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        if min_idx != i:
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr


def bubble_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

#O(n log n)
def merge(arr, l, m, r):
    n1 = m - l + 1
    n2 = r - m

    L = [0] * n1
    R = [0] * n2

    for i in range(n1):
        L[i] = arr[l + i]
    for j in range(n2):
        R[j] = arr[m + 1 + j]

    i = j = 0
    k = l

    while i < n1 and j < n2:
        if L[i] <= R[j]:
            arr[k] = L[i]
            i += 1
        else:
            arr[k] = R[j]
            j += 1
        k += 1

    while i < n1:
        arr[k] = L[i]
        i += 1
        k += 1
    while j < n2:
        arr[k] = R[j]
        j += 1
        k += 1

def mergeSort(arr, l, r):
    if l < r:
        m = l + (r - l) // 2
        mergeSort(arr, l, m)
        mergeSort(arr, m + 1, r)
        merge(arr, l, m, r)

arr = [12, 11, 13, 5, 6, 7]
print("Given array is:", arr)

mergeSort(arr, 0, len(arr) - 1)
print("Sorted array is:", arr)


# Quick sort O(log n)
def quick_sort(arr, low=0, high=None):
    if high is None:
        high = len(arr) - 1
    if low >= high:
        return arr
    p = partition(arr, low, high)
    quick_sort(arr, low, p - 1)
    quick_sort(arr, p + 1, high)
    return arr


def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1
```

Modulo Exponential

```python
def mod_expo(a, n, k):
    if n == 1:
        return a % k

    if n % 2 == 0:
        tmp = mod_expo(a, n // 2, k) #recursive ตัวเองอีกรอบ จนกว่าจะ devide ไม่ได้แล้วจะค่อยๆจบ loop ไปด้านบน
        return (tmp * tmp) % k

    else:
        tmp = mod_expo(a, n // 2, k)
        tmp = (tmp * tmp) % k
        return (tmp * tmp * (a % k)) % k # ---> มาจาก a^x * a^x * a ซึ่ง a คือ a mod k ตัวสุดท้ายที่ devide ต่อไม่ได้แล้ว 
```
