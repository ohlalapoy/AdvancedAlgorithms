```python

def knapsack(n, W, weights, values): #รับค่าเข้ามา มีำจำนวนชิ้น น้ำหนักรวมถุง น้ำหนักย่อยแต่ละชิ้น และมูลค่าของ ของแต่ละขิ้น
    
    # สรา้งตาราง Matrix เก็บค่าของเจ้าตัว สิ่งของและน้ำหนักเข้ามา M[ของจำนวน n ชิ้น][น้ำหนักรวมปัจจุบัน] เริ่มทื่ M[0,w]
    M = [[0] * (W + 1) for _ in range(n + 1)]
    """
    [[0, 0, 0, 0, 0, 0],  ไม่มีของเลย 
    [0, 15, 15, 15, 15, 15], มีของ 1 ชิ้น
    [0, 15, 15, 20, 35, 35],  มีของ 2 ชิ้น
    [0, 15, 15, 20, 35, 45]]  มีของ 3 ชิ้น
    """
  
    for w in range(W + 1):  #ถ้าของในถุง ไม่มีเลย น้ำหนักจะเป็น 0
        M[0][w] = 0
   

    for i in range(1, n + 1): #คิด กรณีมีของเริ่มที่ 1 ชิ้่นเป็นต้นไป
        for w in range(W + 1): #น้ำหนักเริ่มที่ 0 ไปจนถึงน้ำหนักรวมทั้งหทด W
            if weights[i-1] > w: #ถ้าน้ำหนักเกิน
                M[i][w] = M[i-1][w] #ให้ใช้ผลรวมน้ำหนักชิ้นก่อนหน้า ไม่นับชิ้นนี้
            else: #ยังใส่ขอวงได้อีก 
                M[i][w] = max( #เลือก max ระหว่าง
                    M[i-1][w], #ไม่นิ้บชิ้นปัจจุบัน
                    values[i-1] + M[i-1][w-weights[i-1]]) #เอาชิ้นปัจจุบัน อันไหนมากกว่า เลือกอันนั้น 
      
  
      
    selected = [] #เก็บค่า ของที่เอาเอาใส่ถุงมาแล้ว ตอนแรกเป้นถุงว่างเปล่า 
    w = W # น้ำหนักเริ่มต้นเป้น Max เพราะเราจะนับน้ำหนักย้อนไปว่าในถุงมีชิ้นไหนบ้าง 

    for i in range(n, 0, -1): #นับย้อนกลับไป ถอยทีละ 1 
        if M[i][w] != M[i-1][w]:
            selected.append(i)  # หมายเลขชิ้น เริ่มที่ 1
            w -= weights[i-1]  # เมื่อนับย้อน weight จะค่อยๆลดลง 

    selected.reverse() #เรียวข้อมูลกลับมา

    return M[n][W], selected 


weights = [1, 2, 5, 6, 7]
values = [1, 6, 18, 22, 28]
n = len(weights)
W = 11

max_value, selected = knapsack(n, W, weights, values)

print("มุลค่าสิ่งของในถุง สูงสุด:", max_value)
print("เลือกของชิ้นที่ :", selected)
```
